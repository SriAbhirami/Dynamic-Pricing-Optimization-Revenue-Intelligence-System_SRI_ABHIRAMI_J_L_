```python
import pandas as pd

# Load dataset
df = pd.read_csv("../datasets/raw/ecommerce_customer_behavior_dataset_v2.csv")

# Display first 5 rows
df.head()
```

```python
# Shape of the dataset
print("Dataset Shape:", df.shape)

# Column names
print("\nColumns:")
print(df.columns)

# Data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python
# Save processed dataset
df.to_csv(
    "../datasets/processed/ecommerce_customer_behavior_processed.csv",
    index=False
)

print("Processed dataset saved successfully!")
```
